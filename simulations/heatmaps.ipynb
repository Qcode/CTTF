```python
import pandas as pd
import numpy as np
from tqdm import tqdm

map_over_time = np.zeros((48, 200, 200), dtype='int')

print("Getting day 0")
day_0 = df.loc[df['d'] == 0]

print("Grouping by x, y, t")
contact_groups = day_0.groupby(['x', 'y', 't'])['uid'].apply(list)
print("Grouping by uid, t")
pivot_df = day_0.pivot_table(index='uid', columns='t', values=['x', 'y'], aggfunc='first')

originator = 1 #random.randrange(0, 100000)
infected = {originator}

# Spread infection for each time step
for time_step in tqdm(range(48), desc="Processing Time Steps"):
    current_infected = infected.copy()
    # Process each infected user and spread infection
    for user in current_infected:
        x_location = pivot_df.loc[user, ('x', time_step)]
        y_location = pivot_df.loc[user, ('y', time_step)]
        if pd.isna(x_location) or pd.isna(y_location):
            continue
            
        map_over_time[time_step][int(x_location)-1][int(y_location)-1] += 1
        contacts = contact_groups[(x_location, y_location, time_step)]
        for contact in contacts:
            if contact not in infected:
                infected.add(contact)
                    
    print("Time step", time_step, "Infected:", len(infected))

```
