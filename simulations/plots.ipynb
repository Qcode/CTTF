```python
import matplotlib.pyplot as plt
import numpy as np

def print_graph(strategy):
    plt.figure()
    plt.title(f"Average Leech Ranking After One Day, Adversary Strategy = {strategy}")
    plt.ylabel("Ranking")
    plt.xlabel("Page Index")
    for multiplier in (1, 2, 5, 10, 25):
        plt.plot(np.arange(0, 10000), np.load(f"data/leech/overall-a0.05-am{multiplier}-s{strategy}.npy"), label = f"Multiplier = {multiplier}")
    plt.plot(np.arange(0, 10000), np.load(f"data/page-popularity.npy"), label="Ground Truth")
    plt.legend()
    plt.show()

print_graph("OPPOSITE")
print_graph("PIECEWISE")
```
